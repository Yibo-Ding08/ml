```python
"""
- MNIST手写数字识别
- 1999年发布，成为分类算法基准测试的基础
- MNIST仍然是研究人员和学习者的可靠资源

需求
- 从数万个手写图像的数据集中正确识别数字
每张图片都是由28*28像素组成,我们的csv文件中每一行都有784个像素点,表示图片(每个像素)的颜色
"""
```

```python
#导包
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split#分割训练集和测试集
from sklearn.preprocessing import StandardScaler#数据标准化的
from sklearn.neighbors import KNeighborsClassifier#KNN算法,分类对象
from sklearn.metrics import accuracy_score#模型评估,计算模型预测的准确率
import joblib
from pathlib import Path
from collections import Counter

MODEL_PATH = Path("model/手写数字识别.pkl")
```

```python
#定义函数,接受用户传入的索引
def show_digit(idx):
    #读数据
    df = pd.read_csv("../练习数据/手写数字识别.csv")
    #print(df)
    #判断索引是否越界
    if idx <0 or idx>len(df)-1:
        print("索引越界")
        return
    x=df.iloc[:,1:]
    y=df.iloc[:,0]
    #查看用户传入的索引照篇-->是几 
    print(f"该图片对应的数字是:{y.iloc[idx]}")
    print(f"查看所有的标签分布情况:{Counter(y)}")#统计y轴的值
    #查看x的形状
    print(x.iloc[idx].shape)#(784,),想办法变为(28,28)
    #print(x.iloc[idx].values)#具体的784个像素点数据
    #转换为(28,28)
    x=x.iloc[idx].values.reshape(28,28)
    #print(x)
    #具体的绘制
    plt.imshow(x,cmap="grey")#灰度图
    plt.axis("off")
```

```python
def train_model():
    #加载手写数字数据集
    data=pd.read_csv("../练习数据/手写数字识别.csv")
    #print(data)
    #拆分特征列
    x=data.iloc[:,1:]
    y=data.iloc[:,0]
    print(f"x的形状:{x.shape}")#(40000,794)
    print(f"y的形状:{y.shape}")#(40000,)
    print(f"查看y标签分布情况:{Counter(y)}")
    #拆分前进行归一化
    x=x/255
    #拆分训练集和测试集
    x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.8,random_state=24,stratify=y)#随机化标签,参考y轴(数据均衡)
    #创建模型
    estimator=KNeighborsClassifier(n_neighbors=3)
    #训练
    estimator.fit(x_train,y_train)
    #模型评估
    print(f"准确率{estimator.score(x_test,y_test)}")
    print(f"准确率{accuracy_score(y_test,estimator.predict(x_test))}")
    #保存模型
    #参1模型对象,参2,模型保存路径
    MODEL_PATH.parent.mkdir(parents=True, exist_ok=True)
    joblib.dump(estimator, MODEL_PATH)
    print("模型保存成功")
```

```python
def use_model():
    #加载图片
    x=plt.imread("../练习数据/测试数字7.png")#28*28像素
    #绘制图片
    plt.axis("off")
    plt.imshow(x,cmap="grey")
    #plt.show()
    #加载模型
    estimator = joblib.load(MODEL_PATH)
    #模型预测
    #print(x.reshape(1,784).shape)
    print(x.reshape(1,-1).shape)#能转多少转多少
    # plt.imread 读取 PNG 时，像素值已经在 0 到 1 之间
    x = pd.DataFrame(x.reshape(1, -1), columns=estimator.feature_names_in_)
    y_prd=estimator.predict(x)
    print(f"预测值为:{y_prd}")



```

```python
if __name__=="__main__":
    show_digit(9)
    train_model()
    use_model()
```
