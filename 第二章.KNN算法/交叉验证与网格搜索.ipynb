```python
"""
案例:演示网格搜索和交叉验证
交叉验证解释:
     原理:把数据分成n份
     第一次:把第一份当作验证机,后面当作训练集->准确率1
     .......
     计算n次准确率的平均值作为模型最终的准确率
     假设第四次最好,用全部数据训练模型,再用测试机对模型测试
     目的:为了让模型最终验证结果更准确
"""
```

```python
"""
网格搜索:
目的:寻找最优超参数(例如k)
原理:
接受超参数可能出现的值,然后针对于超参的每个值进行交叉验证,获取到最有超参组合
超参数:
需要用户手动录入的数据,不同的超参(组合),可能影响模型的最终评测

"""
```

```python
#导包
from sklearn.datasets import load_iris#加载鸢尾花测试集
from sklearn.model_selection import train_test_split,GridSearchCV#分割训练集和测试集,网格搜索＋交叉验证
from sklearn.preprocessing import StandardScaler#数据标准化的
from sklearn.neighbors import KNeighborsClassifier#KNN算法,分类对象
from sklearn.metrics import accuracy_score#模型评估,计算模型预测的准确率
```

```python
#加载数据集
iris_data=load_iris()
#切分训练集和数据集8:2
x_train,x_test,y_train,y_test=train_test_split(iris_data.data,iris_data.target,test_size=0.2,random_state=12)
#特征工程
transfer=StandardScaler()
x_train=transfer.fit_transform(x_train)
x_test=transfer.transform(x_test)
estimator=KNeighborsClassifier()
#定义字典,记录,超参可能出现的情况(值)
param_dict={"n_neighbors":[i for i in range (1,10)]}
#创建GridSearchCV对象
#参1:模型,参2:该模型超参可能出现的值,4折表示每个超参组合都会进行4折交叉验证,4*10=40次
#返回值:estimator-->处理后的模型对象
estimator=GridSearchCV(estimator,param_dict,cv=4)
estimator.fit(x_train,y_train)
#打印最优超参组合
print(f"最优评分:{estimator.best_score_}")
print(f"最优超参组合:{estimator.best_params_}")
print(f"最优估计器对象:{estimator.best_estimator_}")
print(f"最优交叉验证结果:{estimator.cv_results_}")
```

```python
estimator=estimator.best_estimator_
y_prd=estimator.predict(x_test)
print(y_prd)
```
