```python
#不常用
"数据之间的量纲不同,影响目标结果,因此需要归一化,一般调整到[0,1]之间"
"X'=x-min/max-min"
"x''=x'*(mx-mi)+mi"
"sklearn.preprocessing.MinMaxScaler"
"fit_transform(X)将特征进行归一化处理"
"容易受到最大值和最小值的影响,所以一般用于处理小数据集"
```

```python
from sklearn.preprocessing import MinMaxScaler
x_train=[[90,2,18,40],[60,4,15,45],[75,3,13,46]]
#创建归一化对象
transfer=MinMaxScaler(feature_range=(0,2))
#对原数据进行归一化操作
x_train_new=transfer.fit_transform(x_train)#对应维度转化
print(x_train_new)
```
