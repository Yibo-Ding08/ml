```python
"""
标准化
通过标准化,转化为均值为0标准差为1的标准正态分布的数据
防止应为量纲问题,导致特征列的方差值相差较大,影响模型最终结果
X'=(当前值-该列平均值)/该列标准差
应用场景
    使用于大数据集的处理
"""
```

```python
from sklearn.preprocessing import StandardScaler
x_train=[[90,2,18,40],[60,4,15,45],[75,3,13,46]]
#创建标准化对象
transfer= StandardScaler()
#对原数据集进行标准化操作:
x_train_new=transfer.fit_transform(x_train)
print(x_train_new)
```

```python
#均值
print(transfer.mean_)#下划线,自带属性
print(transfer.var_)#方差
print(transfer.scale_)#标准差
```
